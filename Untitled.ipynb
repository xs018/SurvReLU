```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sksurv.datasets import load_gbsg2, load_veterans_lung_cancer
from sksurv.nonparametric import kaplan_meier_estimator
from sksurv.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sksurv.tree import SurvivalTree
from sklearn import tree
import dtreeviz
```

```python
random_state = 10
X, y = load_gbsg2()

    ## preprocessing
grade_str = X.loc[:, "tgrade"].astype(object).values[:, np.newaxis]
grade_num = OrdinalEncoder(categories=[["I", "II", "III"]]).fit_transform(grade_str)

X_no_grade = X.drop("tgrade", axis=1)
Xt = OneHotEncoder().fit_transform(X_no_grade)
Xt.loc[:, "tgrade"] = grade_num

    # train/test split
X_train, X_test, y_train, y_test = train_test_split(Xt, y, test_size=0.25, random_state=random_state)

st = SurvivalTree(max_depth= 3, min_samples_split=10, min_samples_leaf=15, random_state=random_state)
st.fit(X_train, y_train)

y_train_reg = np.array([float(y[-1]) for y in y_train], dtype=np.float64).reshape(-1, 1)
regr = tree.DecisionTreeRegressor(max_depth=3, random_state=random_state)
regr.fit(X_train, y_train_reg)
```

```python
viz = dtreeviz.model(regr, X_train, y_train_reg)
viz.view(scale=1.0)
```
